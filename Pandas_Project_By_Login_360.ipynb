```python
import pandas as pd
import numpy as np
df = pd.read_csv('/content/medical_insurance_with_claim_date_v2.csv')
df
```

Identify Duplicate Rows

```python
df.duplicated().sum()
```

Remove Duplicate Rows

```python
df.drop_duplicates()
```

Remove Duplicate Claim_ID

```python
df.drop_duplicates(subset="Claim_ID")
```

Remove Duplicate Patient_ID

```python
df.drop_duplicates(subset="Patient_ID")
```

Remove Duplicate Email

```python
df.drop_duplicates(subset="Email")
```

Check Missing Values

```python
df.isnull().sum()
```

Fill Missing Name

```python
df["Name"].fillna("Unknown")
```

Fill Missing Email

```python
df["Email"] = df["Email"].fillna("noemail@example.com")
df["Email"]
```

Fill Missing Hospital Name

```python
df["Hospital_Name"] = df["Hospital_Name"].fillna("Not Available")
df["Hospital_Name"]
```

Fill Missing Insurance Type

```python
df["Insurance_Type"] = df["Insurance_Type"].fillna("General")
df["Insurance_Type"]
```

Fill Missing Claim Amount (Mean by Disease Category)

```python
df["Claim_Amount"] = df["Claim_Amount"].fillna(
    df.groupby("Disease_Category")["Claim_Amount"].transform("mean"))
df["Claim_Amount"]
```

Fill Missing Approved Amount (Median by Insurance Type)

```python
df["Approved_Amount"] = df["Approved_Amount"].fillna(
    df.groupby("Insurance_Type")["Approved_Amount"].transform("median"))
df["Approved_Amount"]
```

```python
df.columns
```

# **Section B — Invalid / Outlier Values**

Remove Claim_Amount < 0

```python
df = df[df["Claim_Amount"] >= 0]
df
```

Remove Approved_Amount > Claim_Amount

```python
df = df[df["Approved_Amount"] <= df["Claim_Amount"]]
df
```

Remove Invalid Age

```python
df = df[(df["Age"] >= 0) & (df["Age"] <= 120)]
df
```

Convert Date Columns

```python
df["Treatment_Start_Date"] = pd.to_datetime(df["Treatment_Start_Date"])
df["Treatment_End_Date"] = pd.to_datetime(df["Treatment_End_Date"])

df["Policy_Start_Date"] = pd.to_datetime(df["Policy_Start_Date"])
df["Policy_End_Date"] = pd.to_datetime(df["Policy_End_Date"])

df['Treatment_Start_Date']
df['Treatment_End_Date']
df['Policy_Start_Date']
df['Policy_End_Date']
```

Remove Invalid Treatment Dates

```python
df = df[df["Treatment_End_Date"] >= df["Treatment_Start_Date"]]
df
```

Remove Invalid Policy Dates

```python
df = df[df["Policy_End_Date"] >= df["Policy_Start_Date"]]
df
```

Fix Gender Typos

```python
df["Gender"].replace({"Mle": "Male","F": "Female"})
```

Fix Payment Mode Typos

```python
df["Payment_Mode"].replace({"Cahs": "Cash"})
```

Standardize Account Status

```python
df["Account_Status"].str.capitalize()
```

Remove Invalid Emails

```python
df[df["Email"].str.contains("@", na=False)]
df[df["Email"].str.contains(".", na=False)]
```

#**Section E — Calculated Columns**

Claim Difference

```python
df["Claim_Difference"] = df["Claim_Amount"] - df["Approved_Amount"]
df["Claim_Difference"]
```

Treatment Duration

```python
df["Treatment_Duration"] = df["Treatment_End_Date"] - df["Treatment_Start_Date"]
df["Treatment_Duration"]
```

Policy Duration

```python
df["Policy_Duration"] = df["Policy_End_Date"] - df["Policy_Start_Date"]
df["Policy_Duration"]
```

Age Group

```python
conditions = [(df['Age'] >= 0) & (df['Age'] <= 17), (df['Age'] >= 18) & (df['Age'] <= 64), (df['Age'] >= 65)]
choices = ['Child', 'Adult', 'Senior']
df['Age_Group'] = np.select(conditions, choices, default='Unknown')
df['Age_Group']
```

Encode Gender

```python
df["Gender_Code"] = df["Gender"].map({"Male": 1,"Female": 0})
df["Gender_Code"]
```

Encode Claim Status

```python
df["Claim_Status_Code"] = df["Claim_Status"].astype("category").cat.codes
df["Claim_Status_Code"]
```
